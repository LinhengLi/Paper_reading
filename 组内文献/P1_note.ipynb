# 卫星互联网系统框架与研究问题定位

## 1. 总体架构

卫星互联网的基本通信路径为：

**用户终端 ↔ 用户链路 ↔ 卫星网络 ↔ 馈线链路 ↔ 网关站 ↔ 地面网络 / Internet**

在这一框架中，用户链路和馈线链路均包含上行与下行两个方向；卫星之间还可以通过星间链路转发数据。

```mermaid
flowchart TB
    U["用户终端<br/>卫星宽带终端 / 支持卫星直连的手机等"]
    S1["接入卫星"]
    S2["其他卫星（可选）"]
    G["网关站 Gateway"]
    N["地面传输网络 / 核心网"]
    I["Internet / 业务服务器"]

    U -->|"用户上行：终端 → 卫星"| S1
    S1 -->|"用户下行：卫星 → 终端<br/>师姐研究：下行波束分配与调度"| U

    G -->|"馈线上行：网关站 → 卫星<br/>你的研究：WRC-27 议题 1.3 相关问题"| S1
    S1 -->|"馈线下行：卫星 → 网关站"| G

    S1 <-->|"星间链路 ISL"| S2
    S2 <-->|"馈线链路（可选路径）"| G

    G <--> N
    N <--> I
```

用户终端不一定是普通手机，也可以是专用卫星宽带终端、车载终端、船载终端或机载终端。普通手机能否直接接入卫星，取决于终端能力及卫星系统设计。

星间链路并非所有系统都具备。具备星间链路时，用户接入的卫星与连接网关站的卫星可以不是同一颗。

## 2. 四类星地通信方向

“用户链路 / 馈线链路”按连接对象划分；“上行 / 下行”按信号在地面与卫星之间的传播方向划分。

| 链路类型 | 传输方向 | 主要作用 |
|---|---|---|
| 用户上行链路 | 用户终端 → 卫星 | 将用户上传的数据、请求等送入卫星网络 |
| 用户下行链路 | 卫星 → 用户终端 | 将业务数据发送至用户，是下行用户波束分配与调度的作用环节 |
| 馈线上行链路 | 网关站 → 卫星 | 将地面网络的数据送入卫星网络 |
| 馈线下行链路 | 卫星 → 网关站 | 将卫星网络的数据送回地面网络 |

用户链路连接卫星与用户终端，馈线链路连接卫星与网关站；两类链路可以分别使用不同的波束及频段。[ESA：多波束卫星通信架构](https://www.esa.int/Enabling_Support/Space_Engineering_Technology/Multiple_Access_Telecom_Reconfigurable_Intersatellites)

## 3. 前向链路与返向链路

“前向 / 返向”描述端到端业务流向，与单段星地链路的“上行 / 下行”不同。

### 3.1 前向链路：地面网络向用户发送数据

典型路径为：

**Internet → 网关站 → 馈线上行 → 卫星 → 用户下行 → 用户终端**

因此，前向链路同时包含：

- 馈线上行链路。
- 用户下行链路。

你研究的馈线上行与师姐研究的用户下行，可以位于同一条前向业务路径中。

### 3.2 返向链路：用户向地面网络发送数据

典型路径为：

**用户终端 → 用户上行 → 卫星 → 馈线下行 → 网关站 → Internet**

因此，返向链路同时包含：

- 用户上行链路。
- 馈线下行链路。

若数据需要跨卫星转发，上述路径还会包含星间链路。前向表示网关侧到用户侧，返向表示用户侧到网关侧。[ESA：前向与返向业务方向](https://resilience.esa.int/archives/projects/folc)

## 4. 你的研究位置：馈线上行链路

如果你此前研究的是 **WRC-27 议题 1.3**，那么将其定位在“馈线上行链路”是准确的。

该议题研究 **51.4–52.4 GHz** 频段用于固定卫星业务（FSS）中网关地球站向非地球静止轨道（NGSO）卫星系统发射的相关问题，即：

**网关地球站 → NGSO 卫星，地对空方向。**

这一范围明确对应网关侧的上行通信。[ITU：WRC-27 议题 1.3 研究文件](https://www.itu.int/md/R23-WP4A-C-0875)

若你的具体工作是该议题下的干扰与兼容性分析，其研究重点可表述为：

> 研究 NGSO 系统网关地球站在指定频段发射时，对相关业务或系统产生的干扰，以及满足共存与保护要求所需的条件。

相关研究包括对邻近频段地球探测卫星业务（无源）的保护，以及多个网关地球站产生的聚合干扰等问题；具体范围仍取决于你研究的干扰场景。[ITU：WRC-23 最后文件，决议 130](https://www.itu.int/dms_pub/itu-r/opb/act/R-ACT-WRC.16-2024-PDF-E.pdf)

## 5. 师姐的研究位置：用户下行链路

按你对师姐工作的描述，其研究位置是：

**卫星 → 地面用户服务小区 / 用户终端。**

核心问题是：

> 面对不同区域随时间变化的业务需求，如何将有限的卫星波束及相关资源分配给地面服务小区？

这里需要区分三个概念：

| 概念 | 含义 |
|---|---|
| 用户终端 | 实际发送或接收业务数据的设备 |
| 服务小区 | 用于覆盖与调度的地面区域，一个小区可以包含多个用户终端 |
| 波束 | 卫星通过天线形成的定向覆盖，是向服务小区提供通信能力的手段 |

因此，“给小区分配波束”通常不等于“给每个用户分配一个独立波束”。

### 5.1 波束跳变调度

波束跳变通过在不同时隙选择不同的服务小区，使有限的波束资源在多个区域之间动态分配。

例如，假设某颗卫星每个时隙最多能同时启用两个用户波束：

| 时隙 | 波束 1 服务的小区 | 波束 2 服务的小区 |
|---|---|---|
| 时隙 1 | 小区 A | 小区 B |
| 时隙 2 | 小区 A | 小区 C |
| 时隙 3 | 小区 D | 小区 B |

若小区 A 的业务需求较高，可以在调度周期内为其安排更多服务时隙。具体调度还可能受到卫星功率、可用波束数量、卫星可见性以及波束间干扰等约束。

### 5.2 Aidos 论文在框架中的位置

根据目前上传的摘要，Aidos 研究的是多卫星、多重覆盖场景中的波束跳变调度，核心任务是快速生成波束跳变时间计划（BHTP），提高吞吐量并降低时延。

结合你描述的用户下行研究场景，可以将其理解为：

> 为卫星面向地面服务小区的波束安排服务时序，使波束资源分配适应不均匀的业务需求，并能在有限的卫星可见时间内完成在线重规划。

具体是否联合优化功率、频率、卫星关联等变量，应以论文后续系统模型与问题定义为准；目前的摘要尚不足以确定全部优化变量。

## 6. 两项研究的联系

| 对比维度 | 你的研究 | 师姐的研究 |
|---|---|---|
| 所在链路 | 馈线上行 | 用户下行 |
| 通信方向 | 网关站 → 卫星 | 卫星 → 用户终端 |
| 主要关注对象 | 网关地球站发射及相关干扰场景 | 用户服务小区与卫星波束资源 |
| 核心问题 | 指定频段的使用与干扰兼容性 | 波束如何按需求分配与调度 |
| 在前向业务路径中的作用 | 将地面数据送入卫星网络 | 将卫星网络中的数据送达用户 |

两项研究在系统层面相互关联：馈线上行提供的数据传输能力，会影响能够送往用户侧的业务量；用户下行的波束调度，则影响这些业务数据如何分配到不同服务区域。

因此，分析端到端性能时，需要同时考虑馈线链路、用户链路，以及存在星间转发时的星间链路容量与时延。
